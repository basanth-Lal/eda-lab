```python
import numpy as np
import pandas as pd
population=pd.Series({'California':78983890,'Texas':84368493,'New York':36536986,'Florida':97655373})
print(population)

df1=pd.DataFrame(population,columns=['population'])
print(df1)

df2=pd.DataFrame(np.random.rand(3,2),columns=['foo','bar'],index=['a','b','c'])
print(df2)

df3=pd.DataFrame([{'a':1,'b':2},{'b':3,'c':4}])
print(df3)

d=[
    {'a':i,'b':2*i}
    for i in range(10)
]
df4=pd.DataFrame(d)
print(df4)
```

```python
import numpy as np
import pandas as pd


data = np.array([
    (1, 'Banuraj', 85),
    (2, 'Rahul', 90),
    (3, 'Kiran', 78),
    (4, 'Arun', 88)
], dtype=[
    ('ID', 'i4'),
    ('Name', 'U10'),
    ('Marks', 'i4')
])


df = pd.DataFrame(data)

print(df)
```

```python
import pandas as pd

ind = pd.Index([2, 3, 5, 7, 11])

print(ind)
print(ind.size)
print(ind.shape)
print(ind.ndim)
print(ind.dtype)
```

```python
print(ind[1])
print(ind[::2])
print(ind[1:4:2])
```

```python
indA=pd.Index([1,3,5,7,9])
indB=pd.Index([2,3,5,7,11])
print(indA & indB)
print(indA | indB)
print(indA ^ indB)
print(indA.intersection(indB))
```
